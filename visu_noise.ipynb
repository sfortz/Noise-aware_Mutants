```python
import pandas as pd
import commons as c
import plotly.express as px
```

# Boxplot of the distance between Ideal and Noisy executions 

```python
type_dict = {
    "Algorithm": str,
    "Qubits_number": int,
    "hardware": str,
    "nature": str,
    "metric": str,
    "ideal_distance": float,
    "noisy_distance": float
}
```

```python
csv_normal_path = 'results_TEST/dataframes/all_data.csv'
df = pd.read_csv(csv_normal_path, dtype=type_dict)

#df = pd.read_csv(csv_normal_path, dtype={col: pd.StringDtype() for col in type_dict.keys()})
```

```python
def print_box_plot(df, file_name, distance):
    
    fig = px.box(df, 
                 x='hardware', 
                 y=distance, 
                 color='nature',  # Définit les couleurs pour chaque valeur de 'nature'
                 title="Boxplot of Distance by noise model and program type",
                 labels={'hardware': "Noise model", distance: "Distance", "nature": "Program type"},
                 boxmode="group")  # Permet de séparer les boxplots par couleur dans chaque groupe
    
    
   # fig.update_layout(boxgroupgap=0, boxgap=0)

    # Save the figure
    output_folder = 'results_TEST/boxplots/'
    c.setup_layout_and_save(fig, "Visualisation of the mutant detectability score for each metric", output_folder, f'{file_name}_{distance}', yaxis_range=[0, 1])
```

```python
for m in c.metrics:
    df_m = df[df['metric'] == m]
    file_name = f'visu_{m}'
    print_box_plot(df_m, file_name, 'noisy_distance')
    print_box_plot(df_m, file_name, 'ideal_distance')
        
```
